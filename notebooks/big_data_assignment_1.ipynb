question 6 

```python
columnas = ["Make", "Model", "Type", "Origin", "DriveTrain",
            "MSRP", "Weight", "Wheelbase", "Length"]

mis_nuevos_datos = mis_datos[columnas][mis_datos["DriveTrain"] == "Front"].copy()

print("Size of mis_nuevos_datos:", mis_nuevos_datos.shape)
```

From the 428 cars, 226 have front wheel drive.

question 7

```python
mis_nuevos_datos.head()
```

```python
mis_nuevos_datos["Weight_kilo"] = mis_nuevos_datos["Weight"] * 0.453592
mis_nuevos_datos["Wheelbase_cm"] = mis_nuevos_datos["Wheelbase"] * 2.54
mis_nuevos_datos["Length_cm"] = mis_nuevos_datos["Length"] * 2.54

mis_nuevos_datos.head()
```

One pound is 0.453592 kilos and one inch is 2.54 centimetres. The average weight is around 1,500 kilos and the average length around 470 centimetres, which is 4.7 metres.

question 8

```python
def opinion(precio):
    if precio > 60000:
        return "Expensive"
    elif precio >= 30000:
        return "Normal"
    else:
        return "Inexpensive"

mis_nuevos_datos["opinion"] = mis_nuevos_datos["MSRP"].apply(opinion)

mis_nuevos_datos["opinion"].value_counts()
```

```python
mis_nuevos_datos.head(10)
```

From the 226 cars with front wheel drive, 175 are Inexpensive, 48 are Normal and only 3 are Expensive.
